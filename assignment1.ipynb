## Q2

```python
products = [i for i in df.columns if i.startswith("ind_") and i.endswith("_ult1")]

df_q2 = df.loc[:, ["fecha_dato", "ncodpers"] + products]
df_q2[products] = df_q2[products].fillna(0).astype("int8")
#고객별로 월 순서대로 정렬해야 바로 위 행이 그 고객의 전월 데이터가 됨.
df_q2 = df_q2.sort_values(["ncodpers", "fecha_dato"]).reset_index(drop=True)
#ncodpers 기준으로 정렬 후 fecha_dato 순으로 정렬 즉, 같은 고객 안에서 날짜 순서대로 정렬함
#새로 정렬한 df에서 인덱스 다시 초기화(reset_index), drop=True: 기존 인덱스 버림
df_q2
```

```python
months = sorted(df_q2["fecha_dato"].unique())
month_idx = df_q2["fecha_dato"].map({m:i for i, m in enumerate(months)})

#shift(1): 데이터 전체를 한 칸 아래로 밀어냄. 즉, i번째 행에 i-1번째 행의 값이 들어감.
#예) 원래 값  [a, b, c, d]
#    shift(1) [0, a, b, c]  <- 맨 위 행은 위에 행이 없어서 빈 칸이 생김
#빈 칸은 기본적으로 NaN이 되지만, fill_value=0을 주면 0으로 채워짐 (int8 타입 유지).
#고객·날짜 순으로 정렬해 두었으므로 prev의 각 행 = 바로 위 행(보통 같은 고객의 전월) 상품 보유 여부.
prev = df_q2[products].shift(1, fill_value=0)
#바로 위 행이 같은 고객이고, 정확히 한 달 전 데이터일 때만 비교 대상으로 사용.
#df_q2["ncodpers"].shift(1): 바로 위 행의 고객 번호 -> 현재 행과 같으면 같은 고객.
#  (고객이 바뀌는 경계에서는 위 행이 다른 고객이라 False가 됨)
#month_idx.shift(1): 바로 위 행의 월 번호 -> 차이가 1이면 정확히 직전 달.
#  (중간에 빠진 달이 있으면 차이가 2 이상이라 False가 됨)
#fill_value 없이 shift하면 첫 행은 NaN이 되고, NaN과의 비교는 항상 False이므로 첫 행도 자동으로 제외됨.
is_prev_month = (df_q2["ncodpers"] == df_q2["ncodpers"].shift(1)) & (month_idx - month_idx.shift(1) == 1)

#이번 달에 보유(1)하고 전월에는 미보유(0)였던 경우 = 신규 가입.
new_products = (df_q2[products] == 1) & (prev == 0)
#위 행이 같은 고객의 직전 달이 아니면 비교가 무의미하므로 신규 가입에서 제외.
new_products.loc[~ is_prev_month] = False # ~ == not
new_products
```

```python
new_products["fecha_dato"] = df_q2["fecha_dato"]
monthly_new = new_products.groupby("fecha_dato")[products].sum()
monthly_new
```

```python
total_new = monthly_new.sum().sort_values(ascending=False)
#전체 기간 동안의 상품별 신규 가입 건수 상위 5개.
top_5 = total_new.head(5)
top_5
```

## Q3

```python
last_month = df.loc[df["fecha_dato"] == "2016-05-28"]
last_month
```

```python
last_month["age_group"] = ["senior" if age >= 65 else ("adult" if age>=30 else "young") for age in last_month["age"]]
```

```python
products = [i for i in last_month.columns if i.startswith("ind_") and i.endswith("_ult1")]

last_month["num_products"] = last_month.loc[:, products].fillna(0).sum(axis=1)
last_month["is_larger_then_3"] = [1 if i >=3 else 0 for i in last_month["num_products"]]
last_month
```

```python
X = last_month.loc[:, ["sexo", "age_group", "segmento", "ind_empleado", "tiprel_1mes"]].dropna()
y = last_month.loc[X.index, "is_larger_then_3"]

X_encoded = pd.get_dummies(X)
X_encoded
```

```python
scores, _ = chi2(X_encoded, y)
scores = pd.Series(scores, index = X_encoded.columns)
scores
```

```python
features = X.columns
features_scores = pd.Series({k:scores[[i for i in X_encoded.columns if i.startswith(k)]].sum() for k in features})
#각 컬럼마다 점수를 하나씩 계산했으므로 하나로 합쳐줌.
features_scores.sort_values(ascending=False)
```

tiprel_mes의 chi squre 가 가장 크므로 이 값이 상품 3개 이상 여부와 가장 관련성이 큰 기준임을 알 수 있다.
